```python
# 0- importando bibliotecas, criando funções e criando listas

#bibliotecas utilizadas
import datetime

#listas de dados dos remédios
remédios = [] #lista para armazenar os nomes dos remédios
dosagem = [] #lista para armazenar as dosagens dos remédios
horário = [] #lista para armazenar os horários dos remédios

#funções
    #função para validar o nome do remédio
def nome_do_remédio():
    while True:
        nome = input("Digite o nome do remédio: ")

        if any(char.isdigit() for char in nome):
            print("Apenas letras!")

        elif nome.lower() in [r.lower() for r in remédios]:
            print("Já existe!")

        else:
            remédios.append(nome)
            print("Adicionado!")
            return nome

    #função para validar a dosagem do remédio
def dosagem_func():
    while True:
        dose = input("Digite a dosagem (mg): ")

        if dose.isdigit():
            dose_int = int(dose)

            if dose_int >= 1000:
                dosagem.append(f"{dose_int/1000} g")
                return f"{dose_int/1000} g"
            else:
                dosagem.append(f"{dose_int} mg")
                return f"{dose_int} mg"

        else:
            print("Apenas números!")
    
    #função para validar o horário do remédio
def horário_func():
    while True:
        hora = input("Digite o horário (HH:MM - 24h): ")

        try:
            datetime.datetime.strptime(hora, "%H:%M")
            horário.append(hora)
            print("Adicionado!")

        except ValueError:
            print("Formato inválido, tente novamente.")
            continue

        opcao = input("Deseja adicionar outro horário? (s/n): ")

        if opcao.lower() != "s":
            break

    return horário.copy()
```

```python
#adicionar remédios, dosagens e horários em um loop

while True:
    nome = nome_do_remédio()
    dose = dosagem_func()
    
    horário.clear() #limpa a lista de horários para cada remédio, evitando acúmulo  de  horários de remédios anteriores
    hora = horário_func()

    with open("lista.txt", "a") as arquivo:
        arquivo.write(f"Remédio: {nome}\n")
        arquivo.write(f"Dosagem: {dose}\n")

        arquivo.write("Horários: " + ", ".join(hora) + "\n")

        arquivo.write("\n")

    opcao = input("Deseja adicionar outro remédio? (s/n): ")

    if opcao.lower() != "s":
        print("Encerrando...")
        break
```

```python

# deletar remédios
nome = input("Digite o nome do remédio que deseja excluir:")

if nome in remédios:
    index = remédios.index(nome)

    del remédios[index]
    del dosagem[index]
    del horário[index]

    # reescreve arquivo inteiro
    with open("lista.txt", "w") as arquivo:
        for i in range(len(remédios)):
            arquivo.write("Remédio: " + remédios[i] + "\n")
            arquivo.write("Dosagem: " + dosagem[i] + "\n")
            arquivo.write("Horário: " + horário[i] + "\n")
            arquivo.write("\n")

    print("Remédio excluído com sucesso!")

else:
    print("Remédio não encontrado.")
```
